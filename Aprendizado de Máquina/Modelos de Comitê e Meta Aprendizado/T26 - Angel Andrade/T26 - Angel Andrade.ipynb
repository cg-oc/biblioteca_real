## Introdução

### Afinal, o que é Aprendizado de Maquina (AM)?

Antes de entender o que é AM, é importante entender o que é Inteligência Artificial (IA), já que AM é somente uma ramificação dessa grande área. Bem, tal termo pode então ser definido, segundo Katti Faceli, em seu livro Inteligência Artificial - Uma Abordagem de Aprendizado de Máquina, como “a ciência e engenharia de criar máquinas inteligentes” [1]. Com isso, diferente do que muitos pensam, IA não é somente os clássicos *ChatBots*, como o *Chat GPT*, na verdade, ela é uma grande área que aborda a tentativa de reproduzir a inteligência humana em uma máquina.

Com isso, AM é uma sub-área da IA, que busca "treinar" algorítmos para induzir modelos que buscam prever algo. Porém, a máquina irá prever o que? E com base em quê? Essas são perguntas centrais para entender como o aprendizado de máquina funciona!

### Ciência de Dados

Para falar de AM, é preciso entender a sua base, dados. Bem, dados podem ser classificados, basicamente, em tabulados ou não tabulados. Dados tabulados tem um tipo de organização baseada em... tabelas! Cada linha significa um exemplo, que apresenta diferentes características que serão a base da análise. Cada coluna, então, representa uma característica, ou *attribute*, do exemplo em si. Já os dados não tabulados se apresentam de maneira diferente, como imagens, textos, etc. Porém aqui só serão usados dados tabulados. Com isso, vamos entender melhor como isso funciona.

#### Tipos de dados

Os dados podem ser divididos em dois grandes grupos:

1. **Dados Numéricos:** tentam expressar de maneira matemática alguma característica. Dados quantitativos podem ser interpretados matemáticamente, permitindo operações entre eles, como normalização, por exemplo.
2. **Dados Categórico:** expressam características em categorias, sem possibilidade de fazer operações matemáticas, exigindo outros tipos de análise.

A partir desses dois tipos, eles são divididos em dois subtipos cada:

**Dados Numéricos:**

* **Dados Contínuos:** se apresentam como números decimais, ou seja, podem apresentar qualquer valor dentro de um certo intervalo. Um exemplo de dado contínuo é altura, pois as pessoas não crescem um número de centímetros definido em um único instante, é um processo *contínuo*.
* **Dados Discretos:** apresentam intervalo definido entre eles, ou seja, se expressam em valores específicos, essencialmente inteiros. Um exemplo de dado discreto é número de irmãos, uma vez que esse é um tipo de dado que só pode ser expresso em números inteiros.

**Dados Categóricos:**

* **Dados Ordinais:** podem ser ordenados em uma escala lógica. Exemplos simples de dados ordinais são aqueles que estão na Escala Likert - que transformam respostas possivelmente abstratas ou subjetivas em uma escala de intensidade, por exemplo a clássica escala de "Discordo totalmente", "Discordo em partes", "Neutro", "Concordo em partes" e "Concordo totalmente".
* **Dados Nominais:** não podem ser ordenados com base em uma ordem lógica, como cor favorita.
* **Dados Binários:**  se apresentam de maneira binária, assumindo apenas dois valores, como macho e fêmea, funcional ou quebrado, etc. Tais dados podem ser mostrados como 0s e 1s, contanto que seja devidamente explicado qual é cada coisa. **Obs.:** Não necessáriamente é possível atribuir ordinalidade a esse tipo de dado, ou seja, 0 e 1 é apenas uma maneira de representar esse dado de maneira mais simples.

Então, ao coletar um conjunto de dados, você tem um *DataSet*, ou seja, um conjunto de dados organizados, que podem ser analisados a partir das suas características e objetivos.

## Como fazer boas previsões?

Certo, digamos que, com esses dados, eu queira prever de que espécie é uma flor com determinados valores de comprimento e largura de suas pétalas. Como fazemos isso? Bem, podemos usar o banco de dados que já temos, tentando encontrar padrões que nos ajudem a entender e prever. Com isso, podemos tentar criar um **modelo** que busque prever o nosso atributo alvo, ou *target*. É para isso que servem os algorítmos de aprendizado de máquina! Mas como fazer isso? Bem, temos várias maneiras, cada uma com suas especificidades. Como nosso caso busca prever um *target* categórico, usamos tipos específicos de algorítmos, como regressão logística e árvore de decisão, pois esse é um problema de **classificação**, ou seja, buscamos um modelo que mostre como classificar dados a partir dos padrões observados.

Mas ainda há uma questão, existem problemas de classificação binária - que buscam prever um *target* binário - e problemas de classificação não binária. Mas a questão mesmo é que classificação não binária é muito mais complexa que a binária, portanto, por enquanto, iremos usar somente dois valores. Como os valores estão igualmente distribuidos, então irei escolher quais eliminar de maneira arbitrária!

```python
filtro = iris["species"] != "setosa"

iris_binario = iris.loc[filtro]

iris_binario
```

Bem, antes de iniciar a indução de um modelo, é preciso dividir os dados, para poder testar a eficiência do modelo. Mas como fazemos isso? Bem, precisamos de uma quantidade de dados para treinar a máquina, para poder induzir ao modelo, e então testar o modelo com exemplos que já sabemos a resposta, com o objetivo de testar a confiabilidade. Ou seja, precisamos de uma divisão em **treino e teste**! Para isso, usaremos a biblioteca ``scikit-learn``.

```python
LEN_TESTE = 0.15

RANDOM_SEED = 3001

iris_index = iris_binario.index

index_treino, index_teste = train_test_split(
    iris_index, test_size=LEN_TESTE, random_state=RANDOM_SEED
)

iris_treino = iris_binario.loc[index_treino]
iris_teste = iris_binario.loc[index_teste]

iris_treino
```

Como podemos ver, por motivos óbvios, nosso dataset de treino é menor que nosso dataset de teste. Isso acontece porque a máquina precisa ter uma quantidade significatica de dados para treinar e induzir o modelo. Ok, agora, antes de continuar precisamos fazer algo que será nossa base de medição da confiabilidade do modelo, uma base line. Mas o que seria uma base line? O modelo mais simples possível, que faça uma previsão sem usar nenhuma informação dos dados de maneira individual. Bem, o modo como escolhemos esse modelo é baseado na melhor maneira de fazer uma previsão somente usando as estatísticas descritivas que temos do nosso atributo *target*, ou seja, no caso de dados categóricos, a moda! Sim, nosso modelo é simplesmente chutar um único valor que siga a moda.

```python
X_treino = iris_treino.reindex(["petal_length", "petal_width"], axis=1)
y_treino = iris_treino["species"]

dummy = DummyClassifier()
dummy.fit(X_treino, y_treino)
```

Agora precisamos de uma maneira de medir quão bom é esse modelo. O principal tipo de medição que usaremos se chama **acurácia**, basicamente, mede quantas vezes o algorítmo errou ou acertou. Para tal, atribuimos cada resposta possível um valor 0 ou 1 - ou, de acordo com a literatura, valor negativo e positivo, respectivamente.

**Obs.:** Não necessariamente valores negativos são inferiores ou ruins em relação aos valores positivos!

A partir disso, podemos definir um critério de avaliação de desempenho a partir das seguintes definições:

* **Verdadeiro Positivo (VP):** O valor predito pelo algorítmo foi um positivo e o valor real era positivo;
* **Falso Positivo (FP):** O valor predito pelo algorítmo foi um positivo, mas o valor real era negativo;
* **Verdadeiro Negativo (VN):** O valor predito pelo algorítmo foi negativo e o valor real era negativo;
* **Falso Negativo (FN):** O valor predito pelo algorítmo foi negativo, mas o valor real.

Com isso, podemos facilmente montar uma tabela que mostra quantos de cada um aconteceram, usando a divisão por treino e teste.

| |Valor predito = 1|Valor predito = 0|
|---|---|---|
| <b>Valor real = 1</b>|VP|FN|
| <b>Valor real = 0</b>|FP|VN|

Assim, para obter a acurácia, basta encontrar o percentual de acerto, da seguinte maneira:

$$
    Acuracia = \frac{VP + VN}{VP + VN + FP + FN}
$$

Além disso, ainda faremos um relatório geral do desempenho dos modelos, mas isso será usada mais adiante.

A biblioteca scikit-learn também tem modos de medir essa acurácia. Vamos testar com nosso base line

```python
X_teste = iris_teste[["petal_length", "petal_width"]]
y_verdadeiro = iris_teste["species"]
ordem_labels = iris_binario["species"].unique().tolist()

y_previsto = dummy.predict(X_teste)

matrix_confusao = confusion_matrix(y_verdadeiro, y_previsto, labels=ordem_labels)

iris_confusao = pd.DataFrame(matrix_confusao, ordem_labels, ordem_labels)

sns.heatmap(iris_confusao, annot=True, annot_kws={"size": 16});
```

Essa é a chamada matriz de confusão, que mostra como de maneira visual o desempenho do algorítmo. Agora vamos começar a usar mais números, vamos medir a acurácia do base line

```python
acuracia_dummy = accuracy_score(y_verdadeiro, y_previsto)

print(f"A acurácia do nosso modelo base-line é de: {(100*acuracia_dummy):.2f}%")
```

Aqui podemos notar um pequeno problema, vamos ver como está a distribuição de espécies com base na divisão de treino e teste que fizemos.

```python
plt.figure(dpi=100)

sns.histplot(
    data=iris_treino,
    x=iris_treino["species"]
)

plt.xlabel("Espécie")
plt.ylabel("Contagem")
```

Como se pode notar, agora há um pequeno desbalanceamento entre as espécies, ou seja, há mais virginicas que versicolors. Como ainda assim os dados estão bem distribuidos, não há tanto problema, pois isso só vai garantir uma moda, mas, em casos de grande desbalanceamento, como, por exemplo, em datasets astronômicos, que tem muito mais estrelas de alguns tipos que de outros, isso é um grande problema!

Bom, mas isso foi uma análise a parte, o que realmente queremos analisar, nesses caso, é nossa acurácia do base-line. Aqui pode-se notar uma coisa, ele teve mais erros que acertos, mas isso se dá por um motivo simples e matemático, como não havia moda nos dados originais, ao dividir em treino e teste, como havia um número ímpar de dados de teste, obrigatóriamente tem uma moda neses data frame e, de maneira análoga, a moda é a outra opção no data frame de treino. Ou seja, a moda dos dados de treino será de uma espécie e, obrigatóriamente, haverão mais dados que não seguem essa moda nos dados de teste, fazendo com que o algorítimo erre mais. ## Árvore de Decisão

Bem, existem muitos algorítmos de classificação, mas um muito famoso é a árvore de decisão, que premite uma melhor interpretação do processo de decisão, já que ele passa por critérios extremamente bem definidos. Vamos dar um exemplo de árvore de decisão por classificação.

```python
PROFUNDIDADE = 5

arvore = DecisionTreeClassifier(criterion="entropy", max_depth=PROFUNDIDADE)
arvore.fit(X_treino, y_treino)

fig, axis = plt.subplots(dpi=200)

plot_tree(
    decision_tree=arvore,
    feature_names=X_treino.columns,
    max_depth=PROFUNDIDADE,
    filled=True
)

plt.show()
```

Certo, agora vamos analisar passo a passo. Primeiro, o que podemos ver é que a árvore de decisão se resume a um grafo unidirecional, onde os vértices calda são as perguntas do algorítmo e os vértices cabeça significam o próximo passo, podendo ser outra pergunta ou uma decisão. Para poder ler essa árvore temos a seguinte "legenda":

* **Vértice raiz (*root node*):** é um vértice cauda, ou pai, que dará a primeira pergunta;
* **Vértice de decisão (*decision nodes*):** é um vértice filho e um vértice pai de outros dois nós, ou seja, ele tem uma seta que vai até ele e duas que partem dele;
* **Vértice folha (*leaf nodes*):** é somente um vértice filho, ou cabeça, ou seja, só há uma seta que vai até ele e nenhuma que parte dele. Esses são os vértices que classificam um dado, caso ele caia nele.

Agora como sabemos se um dado vai para a direita ou para a esquerda? Simples, basta responder a afirmação do nó com verdadeiro ou falso. Caso for verdadeiro, vá pela direita, se for falso, vá pela esquerda! Mas ainda há um último ponto, como definir qual critério de *split* - como são chamadas as divisões geradas pelos nós de decisão, incluindo o nó raiz? Bem, esse é um processo mais complexo que pode envolver vários métodos. Esse, na verdade, é um hiperparâmetro - que veremos logo o que significa - que irá definir como a árvore irá fazer seus *splits*, mas todos eles medem a mesma coisa, o grau de impureza de um determinado nó de decisão, que é o grau de "mistura" de dados de treino que caem em cada uma de suas possíveis respostas.

### Grau de Impureza

Aqui, para medir o quão impuro é um vértice, usaremos o método de da **entropia de Shannon**, que defini a entropia - ou impureza - de um dado vértice como:

$$
    H(X) = -\sum_{x \in X}{p(x)log_{2}(p(x))}
$$

onde $X$ é o conjunto de todos os rótulos no vértice e $p(x)$ é a probabilidade de escolher o rótulo $x$. **Obs.:** Para esse caso, consideraremos $0 \cdot log(0) = 0$, por mais que, matematicamente, essa seja uma indefinição.

Com isso temos a entropia de um vértice, mas o que queremos é, na verdade, a entropia de um *split*! Então, para calculá-la, usaremos como base os dois nós filhos do nosso *split* e definiremos a entropia de um *split* que resulta nos vértices $A$ e $B$ como sendo:

$$
    H(split) = \frac{n_A}{N} H(A) + \frac{n_B}{N} H(B)
$$

onde $n_A$ é o número de exemplos no vértice $A$, $n_B$ o número de exemplos no vértice $B$ e $N$ o total de exemplos que passou pelo split.

### Entendendo a formação da nossa árvore de decisão

Agora, para entender como são definidos esses *splits*, basta definir um critério que minimize a entropia daquele split - vale notar que o método da entropia de Shannon é somente um dos muitos métodos para árvores de classificação. Com isso, inicialmente, analisamos os tipos de atributo que temos.

```python
# Considere o seguinte dataset:

colunas = ["tamanho_pé_(cm)", "altura_(cm)", "idade", "sexo-target"]

data = [
    [16, 125, 10, "feminino"],
    [18, 130, 13, "masculino"],
    [23, 163, 15, "feminino"],
    [26, 174, 17, "masculino"],
    [32, 175, 23, "masculino"],
    [35, 172, 32, "feminino"],
    [30, 176, 25, "masculino"]
]

df = pd.DataFrame(data, columns=colunas)

df
```

Considerando o que se tem, como fazemos a decisão do primeiro split? Certo, primeiro vamos decidir qual atributo usar.

1. Tamanho do pé:

> Agora precisamos analisar as médias entre os vizinhos do dataset, mas primeiro vamos ordená-lo.

```python
tamanho_pe = df.reindex(["tamanho_pé_(cm)"], axis=1).values
tamanho_pe = sorted(tamanho_pe)

testes_split = []

for i in range(len(tamanho_pe)):
    if i < len(tamanho_pe)-1:
        split = (tamanho_pe[i] + tamanho_pe[i + 1])/2
    else:
        split = tamanho_pe[i]
    testes_split.append(float(split[0]))

testes_split
```

Agora eu defino um *split* para cada valor desses e calculo a entropia de cada split, da seguinte maneira:

splits:

1. tamanho_pe_(cm) <= 17.0
2. tamanho_pe_(cm) <= 20.5
3. tamanho_pe_(cm) <= 24.5
.
.
.

Então eu calculo a entropia associada a cada split e escolho aquele com menor entropia. Depois, eu faço o mesmo com os outros dois atributos e escolho aquele que for menor dentre todos. Com isso eu tenho minha primeira escolha, e assim se segue até que os vértices filhos do nó de decisão tenham grau 0 de impureza, caracterizando-o como vértice folha, ou a árvore pare de aprofundar por conta de algum hiperparâmetro.

Agora, caso o atributo seja categórico, é calculada a entropia de cada split referente a um valor único.

### E os tais dos hiperparâmetros?!

Antes mesmo de treinar qualquer algorítmo de aprendizado de máquina, seja ele de classificação ou regressão, é preciso definir seus **hiperparâmetros**, que são parâmetros que irão governar a maneira com a qual o modelo irá se ajustar. Para a árvore de decisão, exemplos são:

* **Critério de split:** qual método será usado para calcular a impureza dos splits;
* **Profundidade:** qual a profundidade máxima da árvore. A profundidade de um vértice se traduz pelo número de arestas até a raiz;
* **Máximo de folhas:** qual o número máximo de folhas que a árvore pode ter;
* **Mínimo de amostras por decisão:** define o número mínimo de amostras para que um vértice seja de decisão. Caso o número seja menor, aquele se torna um vértice folha;
* **Mínimo de amostras por folha:** define um número mínimo de amostras por folha.

A partir da definição desses hiperparâmetros a árvore será construida.

Agora vamos testar, primeiro, a acurácia da nossa primeira árvore, além de medir o quão confiante ela estava para cada predição:

```python
prob_dec = arvore.predict_proba(X_teste)
y_previsto = arvore.predict(X_teste)
acuracia_arvore = accuracy_score(y_verdadeiro, y_previsto)

print(f"A acurácia do nosso modelo por árvore de decisão é de: {100*(acuracia_arvore):.2f}%")

for prob_pred, predict, medicao in zip(prob_dec[:5], y_previsto[:5], y_verdadeiro[:5]):
    print(f" - A flor era uma {medicao}")
    print(f" - A previsão foi de uma {predict}")
    print(f" - A confiança que o modelo atribui à sua escolha foi de {100*prob_pred[ordem_labels.index(predict)]}%")
    print()
```

Ou seja, mesmo que o modelo tenha feito boas previsões, o fato de que ele tenha tido 100% de certeza das suas decisões é algo perigoso, pois a grande acurácia é principalmente graças ao dataset usado, que tem objetivo didático e, por isso, tem grande organização e uma amostra pequena. Porém, na realidade, os dados nem sempre são tão "bonitinhos", podendo ter desbalanceamento, dados anômalos e outros exemplos.

**Obs.:** "confiança" não é um termo tecnicamente correto, essa é somente uma medida da probabilidade atribuida a escolha, advinda da proporção entre a classe escolhida no vértice folha onde o exemplo caiu e o total de exemplos apresentados no treino.

Com isso, podemos classificar as escolhas da seguinte maneira:

* Escolheu corretamente, com alta confiança;
* Escolheu corretamente, com baixa confiança;
* Escolheu errado, com alta confiança;
* Escolheu certo, com baixa confiança.

A partir disso, temos como perceber, por exemplo, que mesmo escolhas erradas tiveram alta confiança, o que é um problema.

### Relatório geral

Agora vamos fazer um relatório geral, usando o módulo ``classification_report`` do *scikit-learn*. Esse relatório mostra diversas estatísticas que tentam avaliar o modelo a partir não só da sua acurácia, mas de medidas de precisão, *recall*, etc.

```python
print(classification_report(
    y_verdadeiro, y_predito, digits=3
))
```

#### precision

> Essa medida é feita para entender o quão preciso o modelo foi nas suas medidas, de maneira mais específica, classe a classe. Ou seja, diferente da acurácia, a precisão vai se perguntar, "das previsões de tumor malígno, quantas acertaram?" e o mesmo para os tumores benignos. Isso ajuda a entender os pontos fracos do modelo. Por exemplo, com dados desbalanceados, mesmo que o modelo induzido por esses dados tenha uma acurácia alta, a precisão da classe menos dominante pode ser zero, pois o modelo deu muito peso para a moda. No caso de tumores, inclusive, isso é um problema seríssimo!

#### recall

> É o contrário da precisão. Ao invés de enteder a precisão das **previsões**, ele busca entender, para cada classe, se os exemplos dessa classe foram previstos corretamente. Ou seja, ele busca responder a "de todos os exemplos de tumor malígno, quantos ele acertou?"

#### f1-score

> Essa é outra medida poderosa, pois resume as medidas de *recall* e precisão, a partir da média harmônica entre elas, ou

$$
    F1_Score = 2 \frac{precision \cdot recall}{precision + recall}
$$

> Essa é uma medida interessante pois busca resumir o desempenho geral para cada tipo de classe do modelo.

#### suport

> Isso só diz quantos exemplos tinham em cada classe. Sim, simples, mas ajuda a entender se havia desbalanceamento de dados e as medidas anteriores. Um complemento, basicamente.

## Viés e Variância

Para entender isso, basta pensar no quão complexo está o modelo. Basicamente, quando deixamos o modelo muito simples - por exemplo uma árvore com 1 ou 2 de profundidade -, podemos fazer com que ele não consiga achar os padrões necessários para fazer suas previsões, pois ele fica simples de mais. Com isso, falamos que o modelo tem **alto viés**, ou *underfitting*, pois ele está altamente enviesado na pouca informação que ele tem. Porém, quando complexificamos demais nosso modelo, colocando uma profundidade muito grande na nossa árvore, por exemplo, o modelo pode encontrar muitos padrões, padrões demais, perdendo a capacidade de generalizar e fazer boas previsões e, nesse caso, dizemos que o modelo tem **alta variância**, ou *overfitting*, pois ele está encontrando padrões onde não há.

No caso da árvore de decisão, ela tem grandes chances de ter *overfitting*, pois são extremamente sensíveis a amostra que estão usando, além de que, com crescimento razoável, ela já acaba encontrando muitos padrões. Por esse motivo tentamos alterar os hiperparâmetros, que geralmente limitam seu crescimento. Por exemplo, um split com poucas amostras *vai* encontrar um padrão, mas poucas amostras não representam bons padrões gerais. Por esse motivo, temos várias estratégias de poda, controle de profundidade, etc., controladas pelos hiperparâmetros.

Mas, para melhorar ainda mais o desempenho dos modelos - não só para árvores de decisão - temos uma estratégia que se basea em democracia, chamada de **aprendizado por comitê**, que une diversos modelos para tentar fazer melhores previsões.

**Obs.:** Algo importante de se deixar claro é que a variância comentada é aplicada ao **modelo** induzido, não aos dados. Os dados em si tem uma medida de variância, usada para entender quanto eles se desviam da média, mas essa é uma variância completamente diferente da usada aqui, mesmo que pareça semelhante. Inclusive, a variância não é uma medida tão usada, por ter unidade de medida quadratica em relação a unidade dos dados. Por esse motivo, a medida de dispersão mais usada nesse caso é o **desvio padrão**, que é de extrema importância para entender como os dados estão distribuidos.

## Floresta Aleatória, usando um comitê de árvores

Um dos modelos de aprendizado por comitê mais famoso é a floresta aleatória, que usa de diferentes árvores, usando *features* e amostras aleatórias para poder ter diferentes decisões e, depois, se basear em todas para ter uma diminuição da variância e do vié de maneira equilibrada. Esse tipo de aprendizado de máquina usa uma estratégia chamada *bagging*, que usa das predições de todos os modelos criados e cada uma equivale a um "voto" para uma predição final. Em problemas de classificação, geralmente a escolha final é baseada na moda dos "votos".

Certo, mas como criar uma floresta boa e próspera? Como as árvores são extremamente sensíveis a amostra que elas usam, usamos uma estratégia de diversificar o máximo possível as árvores, oferecendo diferentes amostras, podendo, inclusive, repetir exemplos, fazendo com que cada árvore tenha uma preferência de atributo, uma sensibilidade diferente. Então agora vamos criar uma floresta!

### Hiperparâmetros

A floresta aleatória tem os mesmos hiperparâmetros de suas árvores - pois também precisamos cuidar das nossas árvores, mesmo que sejam muitas -, mas também tem seus próprios hiperparâmetros. Alguns dos principais são:

* **n_estimators:** número de árvores;
* **bootstrap:** é uma maneira de apresentar as amostras para cada árvore. No caso de ela ser usada, usaremos um número $N$ de amostras - geralmente a amostra de treino - e cada árvore irá usar $N$ amostras, mas a escolha é feita com reposição, ou seja, duas árvores, mesmo que tenham a mesma quantidade de amostras, terão exemplos repetidos diferentes e terão exemplos não usados diferentes.
* **max_features:** porcentagem de atributos que será apresentada para cada split das árvores. Por padrão o valor é 1.0, mas usaremos valores diferentes aqui, para avaliar o desempenho.

**Obs.:** há um parâmetro importante, **n_jobs**, mas que não é um hiperparâmetro. Ele controla quantos núcleos serão usados no computador para criar a floresta, pois, às vezes, ela fica um tanto custosa.

```python
floresta = RandomForestClassifier(criterion="entropy")

floresta.fit(X_treino, y_treino)

y_predito = floresta.predict(X_teste)
prob_pred = floresta.predict_proba(X_treino)

acc = accuracy_score(y_verdadeiro, y_predito)

print(f"A acurácia do modelo foi de {100*(acc):.2f}%")

importancia_atributos = floresta.feature_importances_

ATRIBUTOS = X_treino.columns

eixo = sns.barplot(x=ATRIBUTOS, y=importancia_atributos)
eixo.set_ylabel("Importância relativa")
eixo.set_xlabel("Atributos")
plt.xticks(rotation=90)
plt.show()
```

Aqui podemos notar duas coisas interessantes, primeiro, a acurácia com os valores iniciais dos hiperparâmetros, foi maior que a melhor árvore gerada anteriormente, segundo, percebe-se que alguns atributos tiveram muito mais peso que outros. Com isso, podemos concluir que não precisamos usar todos, até porque, os parâmetros com pouco peso podem, inclusive, sobreajustar o modelo. Então agora vamos testar alterar o hiperparâmetro ``max_features``.

```python
MAX_ATRIBUTOS = [None, 0.7, 0.5, 0.3]

for hiper_prm in MAX_ATRIBUTOS:
    floresta = RandomForestClassifier(criterion="entropy", max_features=hiper_prm)

    floresta.fit(X_treino, y_treino)

    importancia_atributos = floresta.feature_importances_

    print(f"Importância relativa dos atributos, usando um max_feature de {100*hiper_prm}%")

    ATRIBUTOS = X_treino.columns

    plt.figure(dpi=80)
    eixo = sns.barplot(x=ATRIBUTOS, y=importancia_atributos)
    eixo.set_ylabel("Importância relativa")
    eixo.set_xlabel("Atributos")
    plt.xticks(rotation=90)
    plt.show()
```

Como é possível perceber, mesmo variando ``max_features``, percebemos uma maior importância em alguns atributos, como ``worst_area``, ``worst_perimeter``, ``worst_concavity_points``, ``worst_radius`` e ``mean_concavity_points``, o que faz sentido, já que esses parâmetros são ótimos para classificar um tumor, pois isso define, basicamente, seu tamanho e uniformidade. É interessante que o primeiro que fizemos deu importância para vários outros atributos, como ``mean_radius`` e ``mean_texture``, mas os apresentados agora foram os únicos que tiveram grande peso em todos os modelos.

Agora vamos tentar variar outros hiperparâmetros para testar o desempenho.

```python
PROFUNDIDADE = [6, 10, 15]
MIN_SAMPLES_D = [1.0, 10, 20]
MAX_LEAFS = [None, 5, 10]
N_ARVORES = [50, 100, 150]
N_ATRIBUTOS = [0.8, 0.5, 0.3]
RANDOM_SEED = 8971

acuracias = {}

for depth, min_samples, max_leafs, n_trees, n_features in product(
    PROFUNDIDADE,
    MIN_SAMPLES_D,
    MAX_LEAFS,
    N_ARVORES,
    N_ATRIBUTOS
):
    floresta = RandomForestClassifier(
        n_estimators=n_trees,
        criterion="entropy",
        max_depth=depth,
        min_samples_split=min_samples,
        max_leaf_nodes=max_leafs,
        max_features=n_features
    )

    floresta.fit(X_treino, y_treino)
    
    y_predito = floresta.predict(X_teste)
    acc = accuracy_score(y_verdadeiro, y_predito)

    acuracias[acc] = (n_trees, depth, min_samples, max_leafs, n_features)

print(acuracias)
```

Agora vamos usar a floresta com maior acurácia e ver o quão confiante ela estava.

```python
max_acc = max(acuracias.keys())
modelo_vencedor = acuracias[max_acc]


floresta = RandomForestClassifier(
    n_estimators=modelo_vencedor[0],
    criterion="entropy",
    max_depth=modelo_vencedor[1],
    min_samples_split=modelo_vencedor[2],
    max_leaf_nodes=modelo_vencedor[3],
    max_features=modelo_vencedor[4]
)
floresta.fit(X_treino, y_treino)

y_predito = floresta.predict(X_teste)
prob_pred = floresta.predict_proba(X_teste)
classes = [str(classe) for classe in floreta.classes_]

print(f"A acurácia do modelo vencedor foi de {100*(max_acc):.2f}%")

for t_name, p_name, confianca in zip(y_verdadeiro[:9], y_predito[:9], prob_pred[:9]):
    print(f"O tumor era {t_name} e o modelo previu {p_name}, com confiança de {(100*confianca[classes.index(p_name)]):.2f}%")
    print()
```

Como é possível notar, além do aumento da acurácia dos modelos, nós tivemos uma diminuição na *confiança* do modelo com ele mesmo. Isso é muito importante, pois garante o "reconhecimento" das possíveis falhas no mesmo. Isso é reflexo direto do aprendizado por comitê, que, por ter muitas árvores compondo a decisão, apresenta uma menor confiança quando usadas todas as dicisões. Além disso, ela tem grande poder de diminuir a variância do modelo por usar várias árvores com diferentes amostragens de exemplo e atributos. Enquanto isso, uma única árvore, por ser extremamente sensível aos dados apresentados, geralmente apresesnta um grande *overfitting*, principalmente sem estratégias de poda, controle de profundidade ou outras variações dos hiperparâmetros.

Outra coisa interessante de se ressaltar é que, mesmo com o vazamento dos dados, o modelo não pode atribuir grande confiança em muitas de suas escolhas, o que mostra a robustez do aprendizado por comitê, quando se trata de ser realista.
