## Introdução

### Afinal, o que é Aprendizado de Máquina (AM)?

Antes de entender o que é AM, é importante entender o que é Inteligência Artificial (IA), já que AM é somente uma ramificação dessa grande área. Bem, tal termo pode então ser definido, segundo Katti Faceli, em seu livro Inteligência Artificial - Uma Abordagem de Aprendizado de Máquina, como “a ciência e engenharia de criar máquinas inteligentes” [1]. Com isso, diferente do que muitos pensam, IA não é somente os clássicos *ChatBots*, como o *Chat GPT*, na verdade, ela é uma grande área que aborda a tentativa de reproduzir a inteligência humana em uma máquina.

Com isso, AM é uma sub-área da IA, que busca "treinar" algoritmos para induzir modelos que buscam prever algo. Porém, a máquina irá prever o que? E com base em quê? Essas são perguntas centrais para entender como o aprendizado de máquina funciona!

### Ciência de Dados

Para falar de AM, é preciso entender a sua base, dados. Bem, dados podem ser classificados, basicamente, em tabulados ou não tabulados. Dados tabulados tem um tipo de organização baseada em... tabelas! Cada linha significa um exemplo, que apresenta diferentes características que serão a base da análise. Cada coluna, então, representa uma característica, ou *feature*, do exemplo em si. Já os dados não tabulados se apresentam de maneira diferente, como imagens, textos, etc. Porém aqui só serão usados dados tabulados. Com isso, vamos entender melhor como isso funciona.

#### Tipos de dados

Os dados podem ser divididos em dois grandes grupos:

1. **Dados Numéricos:** tentam expressar de maneira matemática alguma característica. Dados quantitativos podem ser interpretados matemáticamente, permitindo operações entre eles, como normalização, por exemplo.
2. **Dados Categórico:** expressam características em categorias, sem possibilidade de fazer operações matemáticas, exigindo outros tipos de análise.

A partir desses dois tipos, eles são divididos em dois subtipos cada:

**Dados Numéricos:**

* **Dados Contínuos:** se apresentam como números decimais, ou seja, podem apresentar qualquer valor dentro de um certo intervalo. Um exemplo de dado contínuo é altura, pois as pessoas não crescem um número de centímetros definido em um único instante, é um processo *contínuo*.
* **Dados Discretos:** apresentam intervalo definido entre eles, ou seja, se expressam em valores específicos, essencialmente inteiros. Um exemplo de dado discreto é número de irmãos, uma vez que esse é um tipo de dado que só pode ser expresso em números inteiros.

**Dados Categóricos:**

* **Dados Ordinais:** podem ser ordenados em uma escala lógica. Exemplos simples de dados ordinais são aqueles que estão na Escala Likert - que transformam respostas possivelmente abstratas ou subjetivas em uma escala de intensidade, por exemplo a clássica escala de "Discordo totalmente", "Discordo em partes", "Neutro", "Concordo em partes" e "Concordo totalmente".
* **Dados Nominais:** não podem ser ordenados com base em uma ordem lógica, como cor favorita.
* **Dados Binários:**  se apresentam de maneira binária, assumindo apenas dois valores, como macho e fêmea, funcional ou quebrado, etc. Tais dados podem ser mostrados como 0s e 1s, contanto que seja devidamente explicado qual é cada coisa. **Obs.:** Não necessáriamente é possível atribuir ordinalidade a esse tipo de dado, ou seja, 0 e 1 é apenas uma maneira de representar esse dado de maneira mais simples.

Então, ao coletar um conjunto de dados, você tem um *DataSet*, ou seja, um conjunto de dados organizados, que podem ser analisados a partir das suas características e objetivos.

## Como fazer boas previsões?

Certo, digamos que, com esses dados, eu queira prever de que espécie é uma flor com determinados valores de comprimento e largura de suas pétalas. Como fazemos isso? Bem, podemos usar o banco de dados que já temos, tentando encontrar padrões que nos ajudem a entender e prever. Com isso, podemos tentar criar um **modelo** que busque prever o nosso atributo alvo, ou *target*. É para isso que servem os algoritmos de aprendizado de máquina! Mas como fazer isso? Bem, temos várias maneiras, cada uma com suas especificidades. Como nosso caso busca prever um *target* categórico, usamos tipos específicos de algoritmos, como regressão logística e árvore de decisão, pois esse é um problema de **classificação**, ou seja, buscamos um modelo que mostre como classificar dados a partir dos padrões observados.

Mas ainda há uma questão, existem problemas de classificação binária - que buscam prever um *target* binário - e problemas de classificação não binária. Mas a questão mesmo é que classificação não binária é muito mais complexa que a binária, portanto, por enquanto, iremos usar somente dois valores. Como os valores estão igualmente distribuidos, então irei escolher quais eliminar de maneira arbitrária!

```python
filtro = iris["species"] != "setosa"

iris_binario = iris.loc[filtro]
```

Bem, antes de iniciar a indução de um modelo, é preciso dividir os dados, para poder testar a eficiência do modelo. Mas como fazemos isso? Bem, precisamos de uma quantidade de dados para treinar a máquina, para poder induzir ao modelo, e então testar o modelo com exemplos que já sabemos a resposta, com o objetivo de testar a confiabilidade. Ou seja, precisamos de uma divisão em **treino e teste**! Para isso, usaremos a biblioteca ``scikit-learn``.

```python
LEN_TESTE = 0.15

iris_index = iris_binario.index

index_treino, index_teste = train_test_split(
    iris_index, test_size=LEN_TESTE, random_state=RANDOM_SEED
)

iris_treino = iris_binario.loc[index_treino]
iris_teste = iris_binario.loc[index_teste]
```

Vale notar que o conjunto de treino é maior que o conjunto de teste. O motivo para isso é que os algoritmos precisam de um grande conjunto de dados para poder encontrar padrões gerais. Ok, agora, antes de continuar precisamos fazer algo que será nossa base de medição da confiabilidade do modelo, uma base line. Mas o que seria uma base line? O modelo mais simples possível, que faça uma previsão sem usar nenhuma informação dos dados de maneira individual. Bem, o modo como escolhemos esse modelo é baseado na melhor maneira de fazer uma previsão somente usando as estatísticas descritivas que temos do nosso atributo *target*, ou seja, no caso de dados categóricos, a moda! Sim, nosso modelo é simplesmente chutar um único valor que siga a moda.

```python
X_treino = iris_treino.reindex(["petal_length", "petal_width"], axis=1)
y_treino = iris_treino["species"]

dummy = DummyClassifier()
dummy.fit(X_treino, y_treino)
```

Agora precisamos de uma maneira de medir quão bom é esse modelo. O principal tipo de medição que usaremos se chama **acurácia**, basicamente, mede quantas vezes o algoritmo acertou. Para tal, atribuimos cada resposta possível um valor 0 ou 1 - ou, de acordo com a literatura, valor negativo e positivo, respectivamente.

**Obs.:** Não necessariamente valores negativos são inferiores ou ruins em relação aos valores positivos!

A partir disso, podemos definir um critério de avaliação de desempenho a partir das seguintes definições:

* **Verdadeiro Positivo (VP):** O valor predito pelo algoritmo foi um positivo e o valor real era positivo;
* **Falso Positivo (FP):** O valor predito pelo algoritmo foi um positivo, mas o valor real era negativo;
* **Verdadeiro Negativo (VN):** O valor predito pelo algoritmo foi negativo e o valor real era negativo;
* **Falso Negativo (FN):** O valor predito pelo algoritmo foi negativo, mas o valor real.

Com isso, podemos facilmente montar uma tabela que mostra quantos de cada um aconteceram, usando a divisão por treino e teste.

| |Valor predito = 1|Valor predito = 0|
|---|---|---|
| <b>Valor real = 1</b>|VP|FN|
| <b>Valor real = 0</b>|FP|VN|

Assim, para obter a acurácia, basta encontrar o percentual de acerto, da seguinte maneira:

$$
    Acuracia = \frac{VP + VN}{VP + VN + FP + FN}
$$

Além disso, ainda faremos um relatório geral do desempenho dos modelos, mas isso será usada mais adiante.

A biblioteca scikit-learn também tem modos de medir essa acurácia. Vamos testar com nosso base line

```python
X_teste = iris_teste[["petal_length", "petal_width"]]
y_verdadeiro = iris_teste["species"]
ordem_labels = iris_binario["species"].unique().tolist()

y_previsto = dummy.predict(X_teste)

matrix_confusao = confusion_matrix(y_verdadeiro, y_previsto, labels=ordem_labels)

iris_confusao = pd.DataFrame(matrix_confusao, ordem_labels, ordem_labels)

sns.heatmap(iris_confusao, annot=True, annot_kws={"size": 16});
```

Essa é a chamada matriz de confusão, que mostra como de maneira visual o desempenho do algoritmo. Agora vamos começar a usar mais números, vamos medir a acurácia do base line

```python
acuracia_dummy = accuracy_score(y_verdadeiro, y_previsto)

print(f"A acurácia do nosso modelo base-line é de: {(100*acuracia_dummy):.2f}%")
```

Aqui podemos notar que, mesmo com distribuição igual de espécies no nosso dataset, o modelo base-line tem acurácia menor que a metade. Por que será? Vamos ver agora nossa distribuição no conjunto de treino, para entender melhor.

```python
plt.figure(dpi=80)

sns.histplot(
    data=iris_treino,
    x=iris_treino["species"]
)

plt.xlabel("Espécie")
plt.ylabel("Contagem")
```

Como se pode notar, agora há um pequeno desbalanceamento entre as espécies, ou seja, há mais virginicas que versicolors. Como ainda assim os dados estão bem distribuidos, não há tanto problema, pois isso só vai garantir uma moda, mas, em casos de grande desbalanceamento, como, por exemplo, em datasets astronômicos, que tem muito mais estrelas de alguns tipos que de outros, isso é um grande problema!

Bom, mas isso foi uma análise a parte, o que realmente queremos analisar, nesses caso, é nossa acurácia do base-line. Aqui pode-se notar uma coisa, ele teve mais erros que acertos, mas isso se dá por um motivo simples e matemático, como não havia moda nos dados originais, ao dividir em treino e teste, como havia um número ímpar de dados de teste, obrigatóriamente tem uma moda neses data frame e, de maneira análoga, a moda é a outra opção no data frame de treino. Ou seja, a moda dos dados de treino será de uma espécie e, obrigatóriamente, haverão mais dados que não seguem essa moda nos dados de teste, fazendo com que o algorítimo erre mais. Mas esse é só o base-line, vamos tentar uma abordagem diferente.

## Árvore de Decisão

Bem, existem muitos algoritmos de classificação, mas um muito famoso é a árvore de decisão, que premite uma melhor interpretação do processo de decisão, já que ele passa por critérios extremamente bem definidos. Vamos dar um exemplo de árvore de decisão por classificação.

```python
PROFUNDIDADE = 5

arvore = DecisionTreeClassifier(criterion="entropy", max_depth=PROFUNDIDADE, random_state=RANDOM_SEED)
arvore.fit(X_treino, y_treino)

fig, axis = plt.subplots(dpi=200)

plot_tree(
    decision_tree=arvore,
    feature_names=X_treino.columns,
    max_depth=PROFUNDIDADE,
    filled=True
)

plt.show()
```

Certo, entender os nós dessa árvore. Primeiro, o que podemos ver é que a árvore de decisão se resume a um grafo unidirecional, onde os vértices calda são as perguntas do algoritmo e os vértices cabeça significam o próximo passo, podendo ser outra pergunta ou uma previsão. Para poder ler essa árvore temos a seguinte "legenda":

* **Vértice raiz (*root node*):** é um vértice cauda, ou pai, que dará a primeira pergunta;
* **Vértice de decisão (*decision nodes*):** é um vértice filho e um vértice pai de outros dois nós, ou seja, ele tem uma seta que vai até ele e duas que partem dele;
* **Vértice folha (*leaf nodes*):** é somente um vértice filho, ou cabeça, ou seja, só há uma seta que vai até ele e nenhuma que parte dele. Esses são os vértices que classificam um dado, caso ele caia nele.

Agora como sabemos se um dado vai para a direita ou para a esquerda? Simples, basta responder a afirmação do nó com verdadeiro ou falso. Caso for verdadeiro, vá pela direita, se for falso, vá pela esquerda! Mas ainda há um último ponto, como definir qual critério de *split* - como são chamadas as divisões geradas pelos nós de decisão, incluindo o nó raiz? Bem, esse é um processo mais complexo que pode envolver vários métodos. Esse, na verdade, é um hiperparâmetro - que veremos logo o que significa - que irá definir como a árvore irá fazer seus *splits*, mas todos eles medem a mesma coisa, o grau de impureza de um determinado nó de decisão, que é o grau de "mistura" de dados de treino que caem em cada uma de suas possíveis respostas.

### E os tais dos hiperparâmetros?!

Antes mesmo de treinar qualquer algoritmo de aprendizado de máquina, seja ele de classificação ou regressão, é preciso definir seus **hiperparâmetros**, que são parâmetros que irão governar a maneira com a qual o modelo irá se ajustar. Para a árvore de decisão, os mais usados são:

* ``criterion``: qual método será usado para calcular a impureza dos splits;
* ``max_depth``: qual a profundidade máxima da árvore. A profundidade de um vértice se traduz pelo número de arestas até a raiz;
* ``max_leaf_nodes``: qual o número máximo de folhas que a árvore pode ter;
* ``min_samples_split``: define o número mínimo de amostras para que um vértice seja de decisão. Caso o número seja menor, aquele se torna um vértice folha;
* ``min_samples_leaf``: define um número mínimo de amostras por folha.

A partir da definição desses hiperparâmetros a árvore será construida.

Agora vamos testar a acurácia da nossa árvore simples:

```python
prob_dec = arvore.predict_proba(X_teste)
y_previsto = arvore.predict(X_teste)
acuracia_arvore = accuracy_score(y_verdadeiro, y_previsto)

print(f"A acurácia do nosso modelo por árvore de decisão é de: {100*(acuracia_arvore):.2f}%")
```

Agora vamos fazer um relatório geral, usando o módulo ``classification_report``, do *scikit-learn*. Esse relatório mostra diversas estatísticas que tentam avaliar o modelo a partir não só da sua acurácia, mas de medidas mais específicas. Isso mostra que, às vezes, uma acurácia boa não significa necessariamente boas previsões para seu objetivo. Para entender melhor isso, vamos ver uma árvore com acurácia de 90%.

```python
print(classification_report(
    y_verdadeiro, y_predito, digits=3
))
```

#### precision

> Essa medida é feita para entender o quão preciso o modelo foi nas suas medidas, de maneira mais específica, classe a classe. Ou seja, diferente da acurácia, a precisão vai se perguntar, "das previsões de tumor malígno, quantas acertaram?" e o mesmo para os tumores benignos. Isso ajuda a entender os pontos fracos do modelo. Por exemplo, com dados desbalanceados, mesmo que o modelo induzido por esses dados tenha uma acurácia alta, a precisão da classe menos dominante pode ser zero, pois o modelo deu muito peso para a moda. No caso de tumores, inclusive, isso é um problema seríssimo!

#### recall

> É o contrário da precisão. Ao invés de enteder a precisão das **previsões**, ele busca entender, para cada classe, se os exemplos dessa classe foram previstos corretamente. Ou seja, ele busca responder a "de todos os exemplos de tumor malígno, quantos ele acertou?"

#### f1-score

> Essa é outra medida poderosa, pois resume as medidas de *recall* e precisão, a partir da média harmônica entre elas, ou

$$
    F1 \ Score = 2 \frac{precision \cdot recall}{precision + recall}
$$

> Essa é uma medida interessante pois busca resumir o desempenho geral para cada tipo de classe do modelo.

#### suport

> Isso só diz quantos exemplos tinham em cada classe. Sim, simples, mas ajuda a entender se havia desbalanceamento de dados e as medidas anteriores. Um complemento, basicamente.

#### macro avg

> A macro average é a média entre os valores das métricas de cada classe, ou

$$
    macro \ avg = \frac{métrica_{malignant} + métrica_{benign}}{2}
$$

#### weighted avg

> A weighted average é a média *ponderada* das métricas, onde os pesos vem de acordo com os valores do *suport*, ajudando a entender o desbalanceamento e o desempenho ponderado do modelo. A expressão que traduz essa medida é

$$
    weighted \ avg = \frac{n_m \cdot métrica_{malignant} + n_b \cdot métrica_{benign}}{n_m + n_b}
$$

### Discutindo o relatório geral

Analisando o relatório gerado, podemos notar que o modelo **realmente** teve um bom desempenho. Isso é ótimo!! Mas vale notar uma coisa, aqui foi usada uma árvore já parametrizada, mesmo que de forma básica. Mas, sem parametrização, essa árvore poderia ter ficado **sobreajustada**, ou seja, ele encontra padrões demais nos dados de treino, perdendo a capacidade de generalizar. Para poder entender isso, usamos o conceito de **viés** e **variância**, que serão explicados a seguir!

## Viés e Variância

Para entender esses conceitos isso, basta pensar no quão complexo está o modelo. Basicamente, quando deixamos o modelo muito simples - por exemplo uma árvore com 1 ou 2 de profundidade -, podemos fazer com que ele não consiga achar os padrões necessários para fazer suas previsões, pois ele fica simples de mais. Com isso, falamos que o modelo tem **alto viés**, ou *underfitting*, pois ele está altamente enviesado na pouca informação que ele tem. Porém, quando complexificamos demais nosso modelo, colocando uma profundidade muito grande na nossa árvore, por exemplo, o modelo pode encontrar muitos padrões, padrões demais, perdendo a capacidade de generalizar e fazer boas previsões e, nesse caso, dizemos que o modelo tem **alta variância**, ou *overfitting*, pois ele está encontrando padrões onde não há.

No caso da árvore de decisão, ela tem grandes chances de ter *overfitting*, pois são extremamente sensíveis a amostra que estão usando, além de que, com crescimento razoável, ela já acaba encontrando muitos padrões. Por esse motivo tentamos alterar os hiperparâmetros, que geralmente limitam seu crescimento. Por exemplo, um split com poucas amostras *vai* encontrar um padrão, mas poucas amostras não representam bons padrões gerais. Por esse motivo, temos várias estratégias de poda, controle de profundidade, etc., controladas pelos hiperparâmetros. Porém, principalmente pelo critério de *split*, podemos ver que mesmo variando os hiperparêmtros, ainda há a capacidade de deixar as folhas - o que definirá suas previsões - puras demais, mas com más previsões.

Mas, para melhorar ainda mais o desempenho dos modelos - não só para árvores de decisão - temos uma estratégia que se basea em união, chamada de **aprendizado por comitê**, que une diversos modelos para tentar fazer melhores previsões.

**Obs.:** Algo importante de se deixar claro é que a variância comentada é aplicada ao **modelo** induzido, não aos dados. Os dados em si tem uma medida de variância, usada para entender quanto eles se desviam da média, mas essa é uma variância completamente diferente da usada aqui, mesmo que pareça semelhante. Inclusive, a variância não é uma medida tão usada, por ter unidade de medida quadratica em relação a unidade dos dados. Por esse motivo, a medida de dispersão mais usada nesse caso é o **desvio padrão**, que é de extrema importância para entender como os dados estão distribuidos.

## A união faz a força - iniciando Aprendizado por Comitê

Para tentar diminuir os problemas dos algorítmos de AM, existem algumas técnicas que usam do ditádo "A união faz a força", as técnicas de aprendizado por comitê. Esses métodos buscam combinar diversos modelos para tentar minimizar os erros dos modelos individuais. Para tal, existem três principais técnicas, o *boosting*, que combina diversos algorítmos em sequência, onde cada um corrige os erros dos outros, buscando melhorar a partir dos erros dos outros, mas ele não será abordado aqui, o *bagging*, que usa um tipo de algorítmo e varia o conjunto de dados apresentados para cada treino, combinando as previsões para melhorar a previsão final - é como se cada previsão individual fosse um voto para a previsão final - e, por fim, o *stacking*, que combina modelos diferentes através de um meta-modelo para fazer suas previsões. Nesse notebook, o foco será sobre *bagging*, além de explicar brevemente o *stacking*, para fins de comparação.

### *Bagging*, a democracia deve nos proteger

#### Dividir para conquistar, usando *bootstrap*

Como explicado acima, a técnica de *bagging* combina diversos modelos de um tipo de algoritmo, treinados com amostras diferentes do conjunto de dados, para poder ter uma maior variabilidade de modelos. Um dos principais algoritmos de *bagging* é a floresta aleatória, que combina diversas árvores para fazer suas previsões. Porém, para variar a amostragem de treino, é preciso usar uma técnica definida e reprodutível que, nesse caso, chama-se ***bootstrap***. Essa técnica define um valor $N$, que geralmente é igual ao tamanho do conjunto de treino, e faz uma amostragem com reposição desse conjunto com $N$ exemplos. Esse tipo de divisão faz com que diferentes amostras possam ter exemplos repetidos, exemplos que faltam em algum treino, mas estão presentes em outros, etc. Isso traz uma grande variabilidade de modelos, principalmente graças a sensibilidade do algorítmo à amostra apresentada para treino. Ou seja, diferentes modelos irão valorizar mais alguns atributos, ou padrões diferentes, decorrente da variação na amostra de treino. Isso faz com que padrões reais e gerais sejam melhor encontrados pelo comitê, onde eles podem ser encontrados em muitos modelos, mesmo variando o modelo.

Para exemplificar, digamos que temos o seguinte conjunto de teste, com 5 exemplos

```python
X = ["A", "B", "C", "D", "E"]
```

A técnica de *bootstrap* irá criar $n$ conjuntos de treino, com 15 exemplos, com reposição, onde $n$ representa a quantidade de modelos. Com isso, teremos

```python
n = 5

bootstrap = [np.random.choice(X, size=len(X)).tolist() for _ in range(n)]
for amostra in bootstrap:
    print(amostra)
    print()
```

Então, como é possível notar, em cada amostra há dados faltantes e dados repetidos, o que faz justamente com que os diferentes modelos tenham grande variação na indução, já que são sensíveis aos dados. Com isso, por exemplo, o modelo $i$ irá dar mais peso para o atributo $a$, enquanto o modelo $j$ dará mais peso para o atributo $b$, pois os dados apresentados foram diferentes.

#### O comitê e seus votos

Depois de dividir os dados, cada modelo é treinado, podendo usar outros tipos de variação, para criar o comitê. Dessa maneira, cada modelo é capaz de fazer uma previsão individual, representando um "voto". Ao final, todos os votos são computados e, se o problema for de regressão - ou seja, númerico - é feita a média dos votos, chegando na previsão final. Caso o problema for de classificação, como é o nosso caso, é computada a moda dos votos. Com isso, padrões gerais são mais facilmente reconhecidos, permitindo menor variância.

Um ponto importante, é que os algoritmos que mais usam o bagging são aqueles que são muito sensíveis e propensos ao *overfitting*, pois o comitê diminui a variância geral do modelo.

## Floresta Aleatória, usando um comitê de árvores

Um dos modelos de aprendizado por comitê mais famoso é a floresta aleatória, que usa de diferentes árvores, usando amostras diferentes de *features* e amostras para poder ter diferentes decisões e, depois, se basear em todas para ter uma diminuição da variância e do vié de maneira equilibrada. Esse é um tipo de aprendizado por *bagging*, usando os votos das árvores individuais para chegar a uma conclusão final. Esse é um processo semelhante ao mostrado anteriormente com os dados sintéticos, mas com maior complexidade, conjunto de hiperparâmetros e maneiras diferentes de se realizar.

**Obs.:** Na verdade, a floresta aleatória não é exatamente um *bagging*, já que usa somente uma amostragem dos atributos para o treino de cada árvore. Mas ele é basicamente um modelo *bagging*.

Certo, mas como criar uma floresta boa e próspera? Como as árvores são extremamente sensíveis a variações nos dados de treino, é utilizada a técnica de *bootstrap* e amostragem de atributos, para gerar a maior variabilidade possível de árvores.

### Hiperparâmetros

A floresta aleatória tem os mesmos hiperparâmetros de suas árvores - pois também precisamos cuidar das nossas árvores, mesmo que sejam muitas -, mas também tem seus próprios hiperparâmetros. Alguns dos principais são:

* **n_estimators:** número de árvores;
* **bootstrap:** se ``True``, as florestas serão treinadas com o *bootstrap*. Se ``False``, a técnica de *bootstrap* não será usada.
* **max_features:** porcentagem de atributos que será apresentada para cada split das árvores. Por padrão o valor é 1.0, mas usaremos valores diferentes aqui, para avaliar o desempenho.

**Obs.:** há um parâmetro importante, **n_jobs**, mas que não é um hiperparâmetro. Ele controla quantos núcleos serão usados no computador para criar a floresta, pois, às vezes, ela fica um tanto custosa. Outra coisa importante é que geralmente não é feito um processo de poda muito complexo, deixando os hiperparâmetros das árvores mais simples, para que as árvores individuais **sejam** mais variadas, permitindo uma maior variabilidade de modelos para buscar uma escolha ótima.

```python
floresta = RandomForestClassifier(
    criterion="entropy",
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=RANDOM_SEED
)

floresta.fit(X_treino, y_treino)

y_predito = floresta.predict(X_teste)
prob_pred = floresta.predict_proba(X_treino)

acc = accuracy_score(y_verdadeiro, y_predito)

print(f"A acurácia do modelo foi de {100*(acc):.2f}%")

importancia_atributos = floresta.feature_importances_

ATRIBUTOS = X_treino.columns

eixo = sns.barplot(x=ATRIBUTOS, y=importancia_atributos)
eixo.set_ylabel("Importância relativa")
eixo.set_xlabel("Atributos")
plt.xticks(rotation=90)
plt.show()
```

Além disso, como será discutido a seguir, não usamos amostragem de atributos, algo característico das florestas.

Depois é criado um vetor ``X_grafico``, com 100 elementos igualmente espaçados no mesmo intervalo de ``X`` para simular um conjunto de teste, para ver como nosso modelo. Por fim, será criada uma floresta com alguns tratamentos que serão discutidos depois e os modelos serão comparados através de gráficos.

```python
# Criando dados sintéticos
from math import pi

rng = np.random.default_rng(3901)       # Gerador de números aleatórios.

X = np.sort(rng.uniform(-pi, pi, 500)).reshape(-1, 1)

y = np.sin(X).ravel() + rng.normal(0, 0.3, size=len(X))

# Pontos para serem previstos
X_grafico = np.linspace(-pi, pi, 700).reshape(-1, 1)

# Criando o comitê de árvores
N_ARVORES = 300
plt.figure(figsize=[7, 5])

floresta = RandomForestRegressor(
    n_estimators=N_ARVORES,
    max_depth=6,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=RANDOM_SEED
    
)
floresta.fit(X, y)
y_predito = floresta.predict(X_grafico)
arvores = floresta.estimators_

for arvore in arvores:
    y_predito_arvore = arvore.predict(X_grafico)
    plt.plot(
        X_grafico,
        y_predito_arvore,
        color="blue",
        alpha=0.5,
        linewidth=0.3
    )

plt.plot(
    X_grafico,
    y_predito,
    color="red",
    linewidth=1.5,
    label="Previsão final"
)
plt.scatter(
    X,
    y,
    color="grey",
    s=10,
    label="Observações"
)

# Criando gaussiana
regressao_real = np.sin(X_grafico)

plt.plot(
    X_grafico,
    regressao_real,
    color="black",
    linewidth=1.5,
    alpha=1,
    label="Regressão real"
)

# fazendo legenda para as árvores
plt.plot(
    [],
    [],
    color="blue",
    alpha=0.5,
    linewidth=0.3,
    label="Previsões das árvores"
)

plt.xlabel("Feature arbitrário")
plt.ylabel("Target arbitrário")
plt.title("Representação do processo de diminuição da variância por bagging.")
plt.legend()
plt.grid(color="gray", linestyle="--", linewidth=0.5, alpha=0.7)
plt.show()
```

Aqui é possível notar como o processo de *bagging* tem uma melhor capacidade de aproximação, onde as linhas azuis representam modelos induzidos por árvores de decisão regressora profundas e a linha vermelha representa o modelo induzido pela média das previsões das árvores, diminuindo a variância do modelo. Algo de se notar é que, ao variar a quantidade de árvores, a variabilidade dos modelos induzidos aumenta junto, mas o modelo induzido pela média se mantém **extremamente estável**, o que mostra o poder do aprendizado por *bagging*.

Algo importante de se notar, é que mantive os hiperparâmetros da árvore anterior, e aumentei a quantidade de árvores. A partir desse tratamento, é possível perceber que, mesmo com um único atributo, a floresta se saiu muito bem, além de que não foi possível variar a amostragem de atributos.

### Com o *stacking*, a meta é o meta-modelo

Acho importante tratar um pouco sobre a ideia inicial de *stacking*, só para não deixar de comentar, mas veremos mais sobre ele mais tarde.

A técnica de *stacking* usa de diversos tipos de modelo diferente, com o mesmo conjunto de treino. Para isso, o *stacking* cria uma hirarquia de níveis, onde modelos de nível 0 - diferentes modelos podem ter o mesmo nível - recebem o conjunto de treino e armazena seus modelos. Com isso, os dados dos modelos de nível 0 são passados para um outro algorítmo de nível 1, e esse processo pode continuar até um modelo melhor. O *stacking* então busca entender as limitações e erros dos modelos de nível inferior, buscando melhor generalização, um **meta-modelo**.

### Entendendo melhor a importância dos atributos para nosso comitê

Agora que entendemos melhor essa técnica e suas vantagens, falta compreender melhor outras duas questões. Primeiro, para deixar as árvores mais variadas, mostramos só uma parte dos atributos. Segundo, ao amostrar os atributos para cada árvore, podemos entender melhor os pesos de cada atributo no treino do modelo.

Bem, para analisar isso, vamos variar a amostragem de atributos e ver o que acontece.

## Um pouco mais de *stacking*

Agora, vamos criar um algoritmo de aprendizado por *stacking*, para ver como ele funciona e fazer algumas medidas de comparação. (**Nota:** Os demais algorítmos usados aqui não serão explicados simplesmente porque serão usados para o *stacking* em específico).

Para isso, usaremos, inicialmente, dois estimadores base:

* Árvore de decisão;
* K-NN.

E, para o meta-modelo, será usado o algoritmo de regressão logística.

```python
modelos_base = [
    ("arvore", DecisionTreeClassifier(random_state=RANDOM_SEED)),
    ("knn", KNeighborsClassifier()),
]

stacking = StackingClassifier(
    estimators=modelos_base,
    final_estimator=LogisticRegression(),
    cv=5,
    stack_method="predict_proba"
)

stacking.fit(X_treino, y_treino)
y_predito = stacking.predict(X_teste)

print(classification_report(y_verdadeiro, y_predito))
```

**Nota:** Aqui, ``estimators`` são os estimadores base - de nível 0 -, ``final_estimator`` é o meta-modelo, induzido, neste caso, por regressão logística, ``cv`` é um parâmetro que define a quantidade amostras para validação cruzada, reduzindo o vazamento de dados, e o ``stack_method`` define as probabilidades dos modelos base - a probabilidade que o modelo dá para sua escolha - como um atributo para o meta-modelo.

Quanto ao relatório, é possível notar uma *redução* das métricas em relação a outros modelos, o que pode ser reflexo da pequena quantidade de modelos base e das limitações dos mesmos. Vamos aumentar a quantidade de modelos, adicionando agora uma floresta aleatória.

```python
arvore = DecisionTreeClassifier(
    criterion="entropy",
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=RANDOM_SEED
)
floresta = RandomForestClassifier(
    criterion="entropy",
    n_estimators=150,
    max_depth=5,
    min_samples_split=10,
    min_samples_leaf=5,
    max_features=0.3,
    random_state=RANDOM_SEED
)

modelos_base = [
    ("arvore", arvore),
    ("floresta", floresta),
    ("knn", KNeighborsClassifier()),
]

stacking = StackingClassifier(
    estimators=modelos_base,
    final_estimator=LogisticRegression(),
    cv=10,
    stack_method="predict_proba"
)

stacking.fit(X_treino, y_treino)
y_predito = stacking.predict(X_teste)

print(classification_report(y_verdadeiro, y_predito))
```

### Grau de Impureza

Aqui, para medir o quão impuro é um vértice, usaremos o método de da **entropia de Shannon**, que defini a entropia - ou impureza - de um dado vértice como:

$$
    H(X) = -\sum_{x \in X}{p(x)log_{2}(p(x))}
$$

onde $X$ é o conjunto de todos os rótulos no vértice e $p(x)$ é a probabilidade de escolher o rótulo $x$. **Obs.:** Para esse caso, consideraremos $0 \cdot log(0) = 0$, por mais que, matematicamente, essa seja uma indefinição.

Com isso temos a entropia de um vértice, mas o que queremos é, na verdade, a entropia de um *split*! Então, para calculá-la, usaremos como base os dois nós filhos do nosso *split* e definiremos a entropia de um *split* que resulta nos vértices $A$ e $B$ como sendo:

$$
    H(split) = \frac{n_A}{N} H(A) + \frac{n_B}{N} H(B)
$$

onde $n_A$ é o número de exemplos no vértice $A$, $n_B$ o número de exemplos no vértice $B$ e $N$ o total de exemplos que passou pelo split.

### Entendendo a formação da nossa árvore de decisão

Agora, para entender como são definidos esses *splits*, basta definir um critério que minimize a entropia daquele split - vale notar que o método da entropia de Shannon é somente um dos muitos métodos para árvores de classificação. Com isso, inicialmente, analisamos os tipos de atributo que temos.

```python
# Considere o seguinte dataset:

colunas = ["tamanho_pé_(cm)", "altura_(cm)", "idade", "sexo-target"]

data = [
    [16, 125, 10, "feminino"],
    [18, 130, 13, "masculino"],
    [23, 163, 15, "feminino"],
    [26, 174, 17, "masculino"],
    [32, 175, 23, "masculino"],
    [35, 172, 32, "feminino"],
    [30, 176, 25, "masculino"]
]

df = pd.DataFrame(data, columns=colunas)

df
```

Considerando o que se tem, como fazemos a decisão do primeiro split? Certo, primeiro vamos decidir qual atributo usar.

1. Tamanho do pé:

> Agora precisamos analisar as médias entre os vizinhos do dataset, mas primeiro vamos ordená-lo.

```python
tamanho_pe = df.reindex(["tamanho_pé_(cm)"], axis=1).values
tamanho_pe = sorted(tamanho_pe)

testes_split = []

for i in range(len(tamanho_pe)):
    if i < len(tamanho_pe)-1:
        split = (tamanho_pe[i] + tamanho_pe[i + 1])/2
    else:
        split = tamanho_pe[i]
    testes_split.append(float(split[0]))

testes_split
```

Agora eu defino um *split* para cada valor desses e calculo a entropia de cada split, da seguinte maneira:

splits:

1. tamanho_pe_(cm) <= 17.0
2. tamanho_pe_(cm) <= 20.5
3. tamanho_pe_(cm) <= 24.5
.

Então eu calculo a entropia associada a cada split e escolho aquele com menor entropia. Depois, eu faço o mesmo com os outros dois atributos e escolho aquele que for menor dentre todos. Com isso eu tenho minha primeira escolha, e assim se segue até que os vértices filhos do nó de decisão tenham grau 0 de impureza, caracterizando-o como vértice folha, ou a árvore pare de aprofundar por conta de algum hiperparâmetro.
